### Gemni

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model1 = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite"
)
```

```python
response1 = model1.invoke("What would you suggest to a first timer to generative AI?Give concise answer in 3 lines.")

print(response1.content)
```

### HuggingFace

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1-0528",
    task="text-generation",
    max_new_tokens=150 # let Hugging Face choose the best provider for you
)
```

```python
model4 = ChatHuggingFace(llm=llm)
```

```python
response4 = model4.invoke("Is there any evidence of God in natural world? Give concise answer in 3 lines.")

print(response4.content)
```

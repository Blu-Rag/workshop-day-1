### We will be using MistralAI model class of langchain for using other parameters

```python
from IPython.display import Markdown, display
```

```python
from langchain_mistralai import ChatMistralAI
```

### Mid Temp

```python
model = ChatMistralAI(
    model_name="ministral-8b-latest",
    temperature=0.7,
    max_tokens=50
)
```

```python
result1 = model.invoke("Write something about how people are realying on AI for everything nowadays.")
```

```python
result1
```

```python
display(Markdown(result1.content))
```
